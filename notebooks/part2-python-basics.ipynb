## Variables: names for values

A **variable** is a name attached to a value. In Python you create one just by assigning with `=` — no announcement of type needed (Java, by contrast, makes you declare `int age = 21;`). Python figures out the type from the value, and the built-in `type()` will tell you what it decided.

```python
age = 21
height = 1.75
name = "Rosa"
is_student = True

print(age, "is a", type(age).__name__)
print(height, "is a", type(height).__name__)
print(name, "is a", type(name).__name__)
print(is_student, "is a", type(is_student).__name__)
```

Those four types — `int` (whole numbers), `float` (numbers with a decimal point), `str` (text, a *string* of characters), and `bool` (`True`/`False`) — cover a huge fraction of everyday programming.

A variable is not glued to its first value or even its first type. Assigning again simply moves the name to a new value:

```python
score = 10
print(score)

score = score + 5      # read the old value, add 5, re-attach the name
print(score)

score = "fifteen"      # legal in Python (Java would refuse!) - but risky style
print(score, "is now a", type(score).__name__)
```

The line `score = score + 5` looks like nonsense algebra, but `=` is not an equation — it means "compute the right side *first*, then attach the name". Changing a variable's type mid-program, as in the last line, is allowed but usually a sign of confusion; good code keeps each name meaning one kind of thing.

## Expressions and arithmetic

An **expression** is any piece of code that produces a value: `2 + 3`, `price * 1.08`, `len(name)`. The arithmetic operators mostly look like a calculator's, with three that surprise newcomers:

| Operator | Meaning | Example |
|---|---|---|
| `/` | division — *always* gives a float | `7 / 2` → `3.5` |
| `//` | floor division — drops the remainder | `7 // 2` → `3` |
| `%` | modulo — *only* the remainder | `7 % 2` → `1` |
| `**` | power | `2 ** 10` → `1024` |

```python
print(7 / 2)        # true division: 3.5
print(7 // 2)       # floor division: 3
print(7 % 2)        # remainder: 1
print(2 ** 10)      # power: 1024

# // and % are a team: quotient and remainder
print(17 // 5, "remainder", 17 % 5)     # 17 = 3*5 + 2
```

## Modulo tricks

That humble `%` operator is secretly one of the most useful tools in programming. Three patterns you will use constantly:

- **even/odd test**: `n % 2` is `0` for even numbers, `1` for odd;
- **last digit**: `n % 10` peels off the final decimal digit;
- **wrap-around**: clocks, weekdays, and circular seating all live on `% 12`, `% 7`, `% n`.

```python
print(42 % 2, 17 % 2)          # 0 = even, 1 = odd
print(1234 % 10)               # last digit of 1234

# Clock arithmetic: 5 hours after 11 o'clock is...
print((11 + 5) % 12)           # 4 o'clock

# divmod does // and % in one call: turn 200 minutes into h + min
hours, minutes = divmod(200, 60)
print(f"200 minutes = {hours} h {minutes} min")
```

## f-strings: readable output

You have already seen `f"..."` strings sneaking into examples. An **f-string** lets you embed any expression inside `{curly braces}`, and add a format spec after a colon: `:.2f` for two decimals, `:,` for thousands separators, `:>8` to right-align in 8 spaces. This beats gluing strings together with `+` in every way.

```python
item = "notebook"
price = 4.5
quantity = 3

print(f"{quantity} x {item} costs {quantity * price:.2f} euros")
print(f"World population is about {8_100_000_000:,} people")

# Alignment makes tidy tables:
for fruit, kcal in [("apple", 52), ("banana", 89), ("avocado", 160)]:
    print(f"{fruit:>10}  {kcal:>4} kcal")
```

## Making decisions: `if` / `elif` / `else`

Programs branch with `if`: a **condition** (an expression that is `True` or `False`) decides which indented block runs. `elif` ("else if") chains extra tests; `else` catches everything left over. Indentation is not decoration in Python — the indented lines *are* the body of the branch.

```python
temperature = 31          # imagine a sensor reported this

if temperature > 35:
    advice = "Heat warning - stay hydrated!"
elif temperature > 25:
    advice = "Warm - t-shirt weather."
elif temperature > 10:
    advice = "Mild - bring a jacket."
else:
    advice = "Cold - wrap up."

print(f"{temperature} degrees: {advice}")
```

Python checks the conditions *top to bottom* and runs only the **first** block whose condition is true — at 31 degrees, both `> 25` and `> 10` are true, but only the "Warm" branch runs. Order matters: test the most specific (here: hottest) case first.

Conditions can be combined with `and`, `or`, and `not`, and comparisons can even be chained the way mathematicians write them:

```python
score = 83                # imagine the user typed this

passed = score >= 60
distinction = score >= 85
print(f"passed: {passed}, distinction: {distinction}")

if 60 <= score < 85:                  # chained comparison - very Pythonic
    print("Solid pass, no distinction.")

is_weekend = False
is_holiday = True
if is_weekend or is_holiday:
    print("No alarm clock tomorrow!")
```

## Repetition: `while` loops

A `while` loop repeats its body *as long as* its condition stays true. It is the right tool when you don't know in advance how many repetitions you need — you only know when to stop.

Classic question: at 7% interest per year, how long until money doubles?

```python
balance = 100.0
years = 0

while balance < 200.0:
    balance = balance * 1.07     # one year of 7% interest
    years += 1                   # shorthand for: years = years + 1

print(f"Doubled after {years} years (balance: {balance:.2f})")
```

The loop ran until the condition `balance < 200.0` became false. The "rule of 72" from finance predicts roughly $72 / 7 \approx 10.3$ years — our exact simulation says 11 full years, matching nicely.

One danger to respect: if the condition *never* becomes false (say you forgot the `years += 1`-style update), the loop runs forever and your notebook hangs. Every `while` loop needs something in its body that pushes it toward stopping.

## Repetition: `for` loops over ranges

When you *do* know how many times to repeat — or want to visit each item of a sequence — use `for`. `range(start, stop, step)` produces a sequence of integers, stopping just *before* `stop` (that half-open convention feels odd at first but makes lengths work out: `range(5)` has exactly 5 items).

```python
for i in range(5):
    print(i, end=" ")            # 0 1 2 3 4  - note: 5 is excluded
print()

for i in range(2, 11, 2):
    print(i, end=" ")            # 2 4 6 8 10
print()

# The accumulator pattern: build a result across iterations
total = 0
for i in range(1, 101):
    total += i
print("1 + 2 + ... + 100 =", total)
```

That last pattern — initialise an **accumulator** before the loop, update it inside, use it after — is the backbone of half of all loops you will ever write. (And the answer, 5050, is the sum young Gauss famously computed by hand.)

Strings are sequences too, so `for` walks them character by character:

```python
sentence = "The quick brown fox jumps over the lazy dog"

vowel_count = 0
for ch in sentence.lower():      # .lower() so 'A' and 'a' both count
    if ch in "aeiou":
        vowel_count += 1

print(f"{vowel_count} vowels in: {sentence!r}")
```

## Functions: packaging steps for reuse

A **function** is a named, reusable block of code: inputs go in as **parameters**, a result comes back via `return`. Once defined, you *call* it as many times as you like. This is how programs stay readable — each function does one job and hides the details.

```python
def area_of_rectangle(width, height):
    return width * height

def greet(name):
    return f"Hello, {name}!"

print(area_of_rectangle(3, 4))
print(area_of_rectangle(2.5, 10))
print(greet("Ada"))
```

Vocabulary check: in the *definition*, `width` and `height` are **parameters** (placeholders). In the *call* `area_of_rectangle(3, 4)`, the values `3` and `4` are **arguments**. The `return` statement hands the result back to whoever called — which is different from `print`, which merely displays text. A function that returns can feed its result into further computation; a function that only prints cannot.

```python
def circle_area(radius, pi=3.14159):     # pi has a DEFAULT value
    return pi * radius ** 2

# Returned values compose - print-only functions wouldn't allow this:
ring = circle_area(5) - circle_area(3)
print(f"Area of a ring with radii 3 and 5: {ring:.2f}")

print(circle_area(1))          # uses the default pi
print(circle_area(1, 3.14))    # overrides it
```

## Drill: FizzBuzz

The famous warm-up: for each number, print `Fizz` if it is divisible by 3, `Buzz` if divisible by 5, `FizzBuzz` if divisible by both, and the number itself otherwise.

The trap is *order*: if you test `% 3 == 0` first, the number 15 gets claimed by `Fizz` and the `FizzBuzz` branch can never run. Most specific case first — the same lesson as the temperature ladder. (Divisible by both 3 and 5 means divisible by 15.)

```python
def fizzbuzz(n):
    if n % 15 == 0:          # both 3 and 5 - MUST be checked first
        return "FizzBuzz"
    if n % 3 == 0:
        return "Fizz"
    if n % 5 == 0:
        return "Buzz"
    return str(n)

for i in range(1, 21):
    print(fizzbuzz(i), end="  ")
print()
```

Notice there is no `elif` here — because each branch `return`s, reaching the next line already implies the earlier tests failed. Both styles are fine; this one reads like a checklist.

## Drill: leap years, with tests

A year is a leap year if it is divisible by 4 — *except* century years, which must be divisible by 400. So 2024: leap. 1900: not (divisible by 100 but not 400). 2000: leap.

This time we will not just eyeball the output — we attach `assert` tests. An `assert` crashes with `AssertionError` if its condition is false and does nothing when it's true, making it a one-line automated test.

```python
def is_leap_year(year):
    if year % 400 == 0:
        return True
    if year % 100 == 0:
        return False
    return year % 4 == 0     # a comparison IS a bool - return it directly

assert is_leap_year(2000) == True
assert is_leap_year(1900) == False
assert is_leap_year(2024) == True
assert is_leap_year(2023) == False
print("All four leap-year tests passed!")
```

Two things to admire in that function. First, the rule ladder runs from most specific (400) to most general (4), so each `return` can be simple. Second, the last line returns the comparison itself — `year % 4 == 0` *is already* `True` or `False`, so wrapping it in `if ... return True else return False` would be pure noise.

## What you just learned

- Variables attach names to values; Python infers types (`int`, `float`, `str`, `bool`) and lets names move.
- `/` vs `//` vs `%` — and modulo's three superpowers: parity, last digit, wrap-around.
- f-strings format anything: `:.2f`, `:,`, `:>10`.
- `if`/`elif`/`else` runs the first true branch — order from specific to general.
- `while` repeats until a condition flips; `for` + `range`/strings visits each item; accumulators build results.
- Functions take parameters, `return` results, and make code reusable and testable with `assert`.
